```python
from transformers import pipeline
```

```python
classifer=pipeline('zero-shot-classification')
```

```python
result = classifer(
    "Ronaldo scores hat trick in Champions League",
    candidate_labels=["sport", "politics", "tech", "business", "entertainment"]
)
print(result)
```

```python
result=classifer(
    "pakistan was about to win the world cup but loses it",
   candidate_labels =['sport','politics','tech','bussiness','entertainment']
)
print(result)
```
